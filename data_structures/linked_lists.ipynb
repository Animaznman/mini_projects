Linked Lists allow for nodes to be built. Each node has a value it holds, as well as a pointer/reference to the next node in its line. Memory is not contiguous, but allows for insert/delete in O(1) if you already know the node.

```python
class Node:
    def __init__(self, value):
        """
        A node stores a value and a pointer to the next node.
        """
        self.value = value
        self.next = None


class LinkedList:
    def __init__(self):
        """
        Initialize an empty linked list.
        Track the head pointer and size.
        """
        self.head = None
        self.size = 0

    def append(self, value):
        """
        Add a new node containing 'value' to the end of the list.
        """
        node = Node(value)
        if self.head:
            current = self.head
            while current.next is not None:
                current = current.next
            current.next = node
        else:
            self.head = node
        
        self.size += 1

    def prepend(self, value):
        """
        Add a new node containing 'value' to the front of the list.
        """
        node = Node(value)
        node.next = self.head
        self.head = node
        
        self.size += 1
        

    def insert(self, index, value):
        """
        Insert a new node containing 'value' at the given index.
        Index 0 means prepend.
        """
        if index < 0 or index >= self.size:
            raise IndexError("Index out of bounds")
        if index == 0:
            self.prepend(value)
            return
    
        node = Node(value)
        current = self.head
        for _ in range(index - 1):
            current = current.next
        node.next = current.next
        current.next = node
        
        self.size += 1

    def delete(self, index):
        """
        Remove the node at the given index.
        Must handle deleting the head separately.
        """
        if index < 0 or index >= self.size:
            raise IndexError("Index out of bounds")
        if index == 0:
            self.head = self.head.next
        else:
            current = self.head
            for _ in range(index - 1):
                current = current.next
            current.next = current.next.next

            self.size -= 1

    def get(self, index):
        """
        Return the value stored at the given index.
        Must traverse the list to reach the node.
        """
        if index < 0 or index >= self.size:
            raise IndexError("Index out of bounds")
        current = self.head
        for _ in range(index):
            current = current.next
        return current.value

    def __len__(self):
        """
        Return the number of nodes in the list.
        """
        return self.size

    def __str__(self):
        """
        Return a readable string representation of the list.
        Example: [10 -> 20 -> 30]
        """
        values = []
        current = self.head
        while current.next is not None:
            values.append(str(current.value))
            current = current.next
        values.append(current.next)
        return "[" + " ->".join(values) + "]"
```
