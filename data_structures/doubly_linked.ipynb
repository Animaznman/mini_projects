Doubly linked lists are similar to single linked lists. However, each nodes has two pointers. A node has a next pointer which points to a node further down the line, and a previous pointer, which is up the line. Doubly linked lists offer bidirectional traversal, fast instert/delete at both **ends**, and easier deletion.

```python
class Node:
    def __init__(self, value):
        """
        A node stores a value and pointers to both the next and previous nodes.
        """
        self.value = value
        self.next = None
        self.prev = None


class DoublyLinkedList:
    def __init__(self):
        """
        Initialize an empty doubly linked list.
        Track head, tail, and size.
        """
        self.head = None
        self.tail = None
        self.size = 0

    def append(self, value):
        """
        Add a new node containing 'value' to the end of the list.
        Update both next and prev pointers.
        """
        
        node = Node(value)
        if self.head is None:
            self.head = node
            self.tail = node
        else:
            self.tail.next = node
            node.prev = self.tail
            self.tail = node
        self.size += 1

    def prepend(self, value):
        """
        Add a new node containing 'value' to the front of the list.
        Update both next and prev pointers.
        """

        node = Node(value)
        if self.head is None:
            self.head = node
            self.tail = node
        else:
            node.next = self.head
            self.head.prev = node
            self.head = node
        self.size += 1


    def insert(self, index, value):
        """
        Insert a new node containing 'value' at the given index.
        Must update both next and prev pointers.
        Index 0 means prepend.
        """
        if index <0 or index > self.size:
            raise IndexError("Index out of bounds.")
        if index == 0:
            self.prepend(value)
            return
        if index == self.size:
            self.append(value)
            return
    
        node = Node(value)
        cur = self.head
        for _ in range(index - 1):
            cur = cur.next
        next = cur.next
        cur.next = node
        node.prev = cur
        node.next = next
        next.prev = node
        self.size += 1
        return 1

    def delete(self, index):
        """
        Remove the node at the given index.
        Must update both next and prev pointers.
        Handle deleting head and tail separately.
        """
        if index <0 or index > self.size:
            raise IndexError("Index out of bounds.")
        if index == 0:
            next = self.head.next
            if next:
                next.prev = None
            
            self.head = next
            if self.head is None:
                self.tail is None
            self.size -= 1
            return
        if index == self.size - 1:
            prev = self.tail.prev
            prev.next = None
            self.tail = prev
            self.size -= 1
            return
        
        if index < 0 and index > -self.size: # Wrap around?
            index = self.size + index
            self.delete(index)
        else:
            if index < self.size // 2:
                cur = self.head
                for _ in range(index):
                    cur = cur.next
                prev = cur.prev
                next = cur.next
                del cur
                prev.next = next
                next.prev = prev
                self.size -= 1
            else:
                cur = self.tail
                for _ in range((self.size - 1) - index):
                    cur = cur.prev
                prev = cur.prev
                next = cur.next
                del cur
                prev.next = next
                next.prev = prev
                self.size -= 1


    def get(self, index):
        """
        Return the value stored at the given index.
        Can traverse from head or tail depending on index for efficiency.
        """
        if index >= self.size or index < 0:
            raise IndexError("Index out of bounds.")
        if index < 0 and index > -self.size: # Wrap around?
            index = self.size + index
            self.get(index)
        if self.head is None and self.tail is None:
            raise LookupError("No nodes exist to return values of.")
        if index < self.size // 2:
            cur = self.head
            for _ in range(index):
                cur = cur.next
        else:
            cur = self.tail
            for _ in range((self.size - 1) - index):
                cur = cur.prev
        
        return cur.value


    def update(self, value, index):
        """
        Bonus I came up with!
        Update the node that exists at the given index and
        overwrite its value.
        """
        if index >= self.size - 1 or index < 0:
            raise IndexError("Index out of bounds.")
        if self.head is None and self.tail is None:
            raise LookupError("No nodes exist to return values of.")
        if index < self.size // 2:
            cur = self.head
            for _ in range(index):
                cur = cur.next
        else:
            cur = self.tail
            for _ in range(index):
                cur = cur.prev
        cur.value = value

    def __len__(self):
        """
        Return the number of nodes in the list.
        """
        return self.size

    def __str__(self):
        """
        Return a readable string representation of the list.
        Example: [10 <-> 20 <-> 30]
        """
        if self.head is None and self.tail is None:
            raise LookupError("Doubly linked list is currently empty.")
        string = "["
        cur = self.head
        for _ in range(self.size - 1):
            string += str(cur.value) + " <-> "
            cur = cur.next
        string += str(cur.value) + "]"
        return string
```
