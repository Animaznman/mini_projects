Hash tables are maps from keys to values. Collision resolution is when multiple keys can map to the same index. Load factors control when a table should resize.

Hash tables provide O(1) average time for insert, delete, and lookup. They're good for counting/frequency tables, deduplication, and indexing.

Chaining collision strategy:
Each bucket stores a linked list of key/value pairs

```python
class HashTable:
    def __init__(self, capacity=8):
        """
        Initialize a hash table with a fixed capacity.
        Uses chaining (lists) to handle collisions.
        """
        self.capacity = capacity
        self.size = 0
        self.buckets = [[] for _ in range(capacity)]

    def _hash(self, key):
        """
        Compute the hash index for the given key.
        Hint:
        - Use Python's built-in hash()
        - Modulo by capacity to stay inside array bounds
        """
        return hash(key) % self.capacity

    def insert(self, key, value):
        """
        Insert or update the key/value pair.

        Hints:
        - Compute the bucket index using _hash(key)
        - Loop through the bucket:
            - If key already exists, update its value
        - If key does NOT exist:
            - Append (key, value) to the bucket
            - Increment size
        - Check load factor:
            - If size / capacity > 0.75, call _resize()
        """
        bucket_index = self._hash(key)
        bucket = self.buckets[bucket_index]
        for index, (k, v) in enumerate(bucket):
            if k == key:
                bucket[index] = (key, value)
                return
        bucket.append((key, value))
        self.size += 1

        if self.size/self.capacity > .75:
            self._resize()
        
    def get(self, key):
        """
        Return the value associated with 'key'.
        Raise KeyError if not found.

        Hints:
        - Compute bucket index
        - Loop through bucket:
            - If key matches, return value
        - If loop ends with no match:
            - Raise KeyError
        """
        bucket_index = self._hash(key)
        bucket = self.buckets[bucket_index]
        for index, (k, v) in enumerate(bucket):
            if k == key:
                return bucket[index][1]
        raise KeyError("Key does not exist in table")

    def delete(self, key):
        """
        Remove the key/value pair.
        Raise KeyError if not found.

        Hints:
        - Compute bucket index
        - Loop through bucket with index:
            - If key matches:
                - Remove the pair
                - Decrement size
                - Return
        - If loop ends with no match:
            - Raise KeyError
        """
        bucket_index = self._hash(key)
        bucket = self.buckets[bucket_index]
        for index, (k, v) in enumerate(bucket):
            if k == key:
                bucket.pop(index)
                self.size -= 1
                return
        raise KeyError("Key does not exist in table")

    def _resize(self):
        """
        Double the capacity and rehash all keys.

        Hints:
        - Save old buckets
        - Double capacity
        - Create new empty buckets
        - Reset size to 0
        - Reinsert every (key, value) pair using insert()
          (this ensures they get rehashed correctly)
        """
        old_buckets = self.buckets
        self.capacity *= 2
        self.buckets = [[] for _ in range(self.capacity)]
        self.size = 0
        for old_bucket in old_buckets:
            for item in old_bucket:
                self.insert(item[0], item[1])


    def __len__(self):
        """
        Return the number of key/value pairs.
        """
        return self.size

    def __str__(self):
        """
        Return a readable representation of the hash table.
        """
        return str(self.buckets)
```
