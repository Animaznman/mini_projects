Graphs are sets of nodes and edges.

Graphs can either be directed or undirected.

Graphs allow for arbitrary relationships of any kind, but have harder algorithms to implement.

Graphs are usually represented in two flavors:
1. Adjacency list
2. Adjacency matrix

Adjacency list: A dictionary mapping each node -> list of neighbors
```
A: [B, C]
B: [C]
C: []
```

Adjacency matrix: 2D grid of 0/1 or weights.

````python
class Graph:
    def __init__(self):
        """
        Store graph as an adjacency list.
        Hint:
        - Use a dictionary: {node: [neighbors]}
        """
        self.graph = {}

    def add_node(self, node):
        """
        Add a node to the graph.
        Hint:
        - If node not in adjacency list, create empty list.
        """
        if node not in self.graph:
            self.graph[node] = []

    def add_edge(self, src, dest):
        """
        Add an edge from src to dest.
        Hint:
        - Ensure both nodes exist.
        - Append dest to src's adjacency list.
        - For undirected graphs, also append src to dest.
        """
        if src in self.graph and dest in self.graph:
            if dest not in self.graph[src]:
                self.graph[src].append(dest)
                return
        # Undirected would be similar code, just slightly reversed
        raise KeyError("Source or destination not in graph")
    def remove_edge(self, src, dest):
        """
        Remove an edge.
        Hint:
        - Remove dest from src's adjacency list.
        - For undirected graphs, also remove src from dest.
        """
        if src in self.graph.keys and dest in self.graph.keys:
            if dest in self.graph[src]:
                self.graph[src].remove(dest)
                return
            else:
                raise KeyError("Destination not in source adjacency")
        raise KeyError("Source or destination not in graph")

    def remove_node(self, node):
        """
        Remove a node and all edges pointing to it.
        Hint:
        - Delete node from adjacency list.
        - Remove node from all neighbor lists.
        """
        if node not in self.graph.keys:
            raise KeyError("Node not in graph")
        for neighbors in self.graph.values():
            if node in neighbors:
                neighbors.remove(node)
        del self.graph[node]

    def bfs(self, start):
        """
        Breadth-first search.
        Hint:
        - Use a queue.
        - Track visited nodes.
        """
        if not self.graph:
            raise KeyError("Graph is empty")
        if start not in self.graph:
            raise KeyError("Start node not in graph")
        queue = [start]
        traveled = []

        while queue:
            popped = queue.pop(0)
            if popped in traveled:
                continue

            traveled.append(popped)

            queue.extend(self.graph[popped])
        
        return traveled


    def dfs(self, start):
        """
        Depth-first search.
        Hint:
        - Use recursion OR a stack.
        - Track visited nodes.
        """
        stack = [start]
        traveled = []
        while stack:
            node = stack.pop()

            if node not in traveled:
                traveled.append(node)

                for neighbor in self.graph[node]:
                    stack.append(neighbor)
        
        return traveled

    def __str__(self):
        """
        Return adjacency list as a string.
        """
        return str(self.graph)
````
