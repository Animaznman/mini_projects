Binary Search Trees are node-based trees that have a left subtree and a right subtree. The boast efficient searches, insertion, and deletion.

```python
class Node:
    def __init__(self, value):
        """
        A BST node stores a value and pointers to left and right children.
        """
        self.value = value
        self.left = None
        self.right = None


class BinarySearchTree:
    def __init__(self):
        """
        Initialize an empty BST.
        Track the root node.
        """
        self.root = None

    def insert(self, value):
        """
        Insert 'value' into the BST.
        Must preserve BST ordering rules.
        """
        node = Node(value)
        if self.root is None:
            self.root = node
        else:
            cur_node = self.root
            while cur_node.left is not None or cur_node.right is not None:
                if value < cur_node.value:
                    if cur_node.left is None:
                        cur_node.left = node
                        return
                    else:
                        cur_node = cur_node.left
                        continue
                elif value > cur_node.value:
                    if cur_node.right is None:
                        cur_node.right = node
                        return
                    else:
                        cur_node = cur_node.right
                        continue
                else:
                    raise ValueError("Value already exists in BST")
                

    def search(self, value):
        """
        Return True if 'value' exists in the BST.
        Otherwise return False.
        """
        cur_node = self.root
        while cur_node:
            if cur_node.value == value:
                return True
            elif value < cur_node.value:
                cur_node = cur_node.left
            elif value > cur_node.value:
                cur_node = cur_node.right
        return False

    def delete(self, value):
        self.root = self._delete(self.root, value)
    
    def _delete(self, node, value):
        if node is None:
            return None
        
        # Searching for node
        if value < node.value:
            node.left = self._delete(node.left, value)
        if value > node.value:
            node.right = self._delete(node.right, vlue)
        
        # No children
        if node.left is None and node.right is None:
            return None
        
        # One child
        if node.left is None:
            return node.right
        if node.right is None:
            return node.left
        
        # Two children
        successor = self._find_min(node.right)
        node.value = successor.value

        node.value = successor.value

        # Delete the successor
        node.right = self._delete(node.right, successor.value)

        return node
    
    def _find_min(self, node):
        while node.left:
            node = node.left
        return node
            
    def find_min(self):
        """
        Return the minimum value stored in the BST.
        Must traverse left children only.
        """
        cur_node = self.root
        while cur_node.left:
            cur_node = cur_node.left
        return cur_node

    def find_max(self):
        """
        Return the maximum value stored in the BST.
        Must traverse right children only.
        """
        cur_node = self.root
        while cur_node.right:
            cur_node = cur_node.right
        return cur_node

    def inorder(self):
        """
        Return a list of values in sorted order.
        (Left, Root, Right)
        """
        traveled = []
        def travel(node):
            if node is None:
                return
            travel(node.left)
            traveled.append(node.value)
            travel(node.right)
        travel(self.root)
        return traveled

    def preorder(self):
        """
        Return a list of values in preorder.
        (Root, Left, Right)
        """
        traveled = []
        
        def travel(node):
            if node is None:
                return
            traveled.append(node.value)
            travel(node.left)
            travel(node.right)
        travel(self.root)
        return traveled

    def postorder(self):
        """
        Return a list of values in postorder.
        (Left, Right, Root)
        """
        traveled = []
        def travel(node):
            if node is None:
                return
            travel(node.left)
            travel(node.right)
            traveled.append(node.value)
        travel(self.root)
        return traveled
    
    def bfs(self):
        """
        Return a level-order Traversal
        """
        if self.root is None:
            return []
        queue = [self.root]
        traveled = []

        while queue:
            node = queue.pop(0)
            traveled.append(node.value)

            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        
        return traveled

    def height(self):
        """
        Return the height of the BST.
        Height of empty tree is -1.
        Height of single-node tree is 0.
        """
        def calculate_height(node):
            if node is None:
                return -1
            left_height = calculate_height(node.left)
            right_height = calculate_height(node.right)
            
            return 1 + max(left_height, right_height)
        return calculate_height(self.root)

    def __str__(self):
        """
        Return a readable string representation of the BST.
        Typically an inorder traversal.
        """
        return str(self.inorder())
```

### Some notes about traversals:

#### Inorder = Left -> Root -> Right
It sorts the tree. Inorder traversal of a BST always produces sorted values.

#### Preorder = Root -> Left -> Right
Preorder is a very top down approach. It allows serialization of trees, making it easier to copy it.

#### Postorder = Left -> Right -> Root
Postorder is helpful for deleting trees from the children all the way to the root.

#### Breadth First Search = Roots before children
Helps print the tree level by level and seeing if the tree is 'complete'.
