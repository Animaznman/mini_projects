Stacks are dynamic linear data structures. They allow for the size of the array to change appropriately.

```python
class Stack:
    def __init__(self, capacity):
        """
        Initialize the stack with a fixed capacity.
        Create internal storage and a pointer for the top element.
        """
        self.capacity = capacity
        self.size = 0
        self.stack = [None] * capacity
    
    def push(self, value):
        """
        Add a value to the top of the stack.
        If the stack is full, raise an error.
        """
        if self.size == self.capacity:
            raise OverflowError("Stack is full")
        self.stack[self.size] = value
        self.size += 1

    def pop(self):
        """
        Remove and return the top element of the stack.
        If the stack is empty, raise an error.
        """
        if self.size <= 0:
            raise IndexError("Stack is empty")
        value = self.stack[self.size - 1]
        self.stack[self.storage - 1] = None
        self.size -= 1

        return value

    def peek(self):
        """
        Return the top element without removing it.
        If the stack is empty, raise an error.
        """
        if self.size <= 0:
            raise IndexError("Stack is empty")
        
        return self.stack[self.size - 1]

    def is_empty(self):
        """
        Return True if the stack has no elements.
        """
        return self.size == 0

    def is_full(self):
        """
        Return True if the stack has reached its capacity.
        """
        return self.size == self.capacity

    def __len__(self):
        """
        Return the number of elements currently in the stack.
        """
        return self.size

    def __str__(self):
        """
        Return a readable string representation of the stack.
        Show elements from bottom to top.
        """
        return "[" + ",".join(str(self.stack[i]) for i in range(self.size)) + "]"
```
