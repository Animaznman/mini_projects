Queues are FIFO data structures. This involves having a head pointer and a tail pointer.

```python
class Queue:
    def __init__(self, capacity):
        """
        Initialize the queue with a fixed capacity.
        Create internal storage, head/tail pointers, and size counter.
        """
        self.capacity = capacity
        self.size = 0
        self.queue = [None] * capacity
        self.head, self.tail = 0, 0

    def enqueue(self, value):
        """
        Add a value to the back of the queue.
        If the queue is full, raise an error.
        Use circular indexing for the tail pointer.
        """
        if self.size == self.capacity:
            raise IndexError("Queue is full")
        self.tail = (self.head + self.size) % self.capacity
        self.queue[self.tail] = value
        self.size += 1
        

    def dequeue(self):
        """
        Remove and return the front element of the queue.
        If the queue is empty, raise an error.
        Use circular indexing for the head pointer.
        """
        if self.size == 0:
            raise IndexError("Queue is empty")
        value = self.queue[self.head]
        self.head = (self.head + 1) % self.capacity
        self.size -= 1

        return value

    def peek(self):
        """
        Return the front element without removing it.
        If the queue is empty, raise an error.
        """
        if self.size == 0:
            raise IndexError("Queue is empty")
        return self.queue[self.head]

    def is_empty(self):
        """
        Return True if the queue has no elements.
        """
        return self.size == 0

    def is_full(self):
        """
        Return True if the queue has reached its capacity.
        """
        return self.size == self.capacity

    def __len__(self):
        """
        Return the number of elements currently stored in the queue.
        """
        return self.size

    def __str__(self):
        """
        Return a readable string representation of the queue.
        Show elements from front to back in logical order.
        """
        items = []
        for i in range(self.size):
            index = (self.head + i) % self.capacity
            items.append(str(self.queue[index]))
        return "[" + ", ".join(items) + "]"
```
