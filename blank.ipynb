```python
def draw_line(x_1, y_1, x_2, y_2):
    coordinates = []
    if x_2 != x_1:
        slope = int(((y_1 - y_2)/(x_1-x_2))) # Watch out for divide by 0
        # (1,1), (5,5)
        print(f"slope: {slope}")
        
        # for x in range(x_1, (x_2 + 1) * 2):
        #     y = int(x*.5) * slope
        #     coordinates.append((x,y))
        
        start = x_1
        while start < x_2:
            y = start * slope
            coordinates.append((int(start),int(y)))
            start += .5
    else:
        for y in range(y_1, y_2 + 1):
            coordinates.append((x_1, y))
        
        

    print(f"coordinates: {coordinates}")
    for coordinate in coordinates:
        x, y = coordinate
        x_man = x
        y_man = len(canvas) - y -1
        print(f"x_man: {x_man}")
        print(f"y_man: {y_man}")
        canvas[y_man][x_man] = "*"

draw_line(2, 2, 5, 9)
canvas
```
